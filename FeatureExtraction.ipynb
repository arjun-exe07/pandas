```python
def extract_episode(txt : str)-> str:
  check = False
  data = ""
  for i in txt:
    if i == ')':
      break
    elif i == '(':
      check = True
    elif check == True:
      data += i
  return data
```

```python
# Make a new column for episode count
df['Episodes'] =df['Title'].apply(extract_episode)
```

```python
df['Episodes'] = df['Episodes'].str.replace(' eps' , '')
```

### Make a new column for time stamp

```python
df.loc[1 ,'Title']
```

```python
#same as above
def extraction_time(txt : str)-> str:
  check = False
  data = ""

  for i in range(len(txt)):
    if txt[i] == ')':
      for j in range(i+1 , i+20):
        data += txt[j]

      return data
```

```python
df['Total Time'] = df['Title'].apply(extraction_time)
```

### Create a Column for total month

```python
from datetime import datetime
from dateutil.relativedelta import relativedelta

def calculate_total_months(period : str) -> int :
  try:
    start , end = period.split(' - ')
    start_date = datetime.strptime(start , '%b %Y')
    end_date = datetime.strptime(end , '%b %Y')

    r = relativedelta(end_date , start_date)
    return r.years * 12 + r.months * 1  # +1 to include the starting month
  
  except:
    return None

df['Months'] = df['Total Time'].apply(calculate_total_months)
```
