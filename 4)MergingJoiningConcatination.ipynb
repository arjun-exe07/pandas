```python
import numpy as np
import pandas as pd
```

### Merging 2 DataFrames

```python
employees_data = {
    'emp_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'department': ['HR', 'IT', 'Sales', 'IT', 'Marketing']
}

salaries_data = {
    'emp_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 75000, 90000, 65000],
    'bonus': [5000, 10000, 7000, 8000, 4000]
}

employees = pd.DataFrame(employees_data)
salaries = pd.DataFrame(salaries_data)
```

```python
employees
```

```python
salaries
```

```python
pd.merge(employees , salaries , on = 'emp_id' ,how="inner")
# how = "inner" is default i.e common on both emp and sal
```

```python
pd.merge(employees , salaries , on = 'emp_id' ,how="outer")
# how = "outer" i.e. it is for all
```

```python
pd.merge(employees , salaries , on = 'emp_id' ,how="left")
```

```python
pd.merge(employees , salaries , on = 'emp_id' ,how="right")
```

### Concatination of 2 DataFrame

```python

df1 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2'],
    'C': ['C0', 'C1', 'C2']
})


df2 = pd.DataFrame({
    'A': ['A3', 'A4', 'A5'],
    'B': ['B3', 'B4', 'B5'],
    'C': ['C3', 'C4', 'C5']
})
```

```python
pd.concat([df1 , df2])  ## on the basis or rows i.e axis = 0
```

```python
pd.concat([df1 , df2] , axis =1 )
```

### Joining 2 Data Frames 

```python
df1 = pd.DataFrame(
    {'name': ['Alice', 'Bob', 'Charlie']},
    index=[1, 2, 3]
)

df2 = pd.DataFrame(
    {'score': [85, 90, 75]},
    index=[2, 3, 4]
)
```

```python
df1
```

```python
df2
```

```python
df1.join(df2 , how = 'outer')
```
