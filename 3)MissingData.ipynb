```python
import numpy as np
import pandas as pd
```

### Finding Missing Data

```python
data = {
  'A': [1 , 2 , np.nan ,4 , 5 ],
  'B': [np.nan , 2, 3, 4, 5],
  'C': [1 ,2 ,3 , np.nan , np.nan],
  'D': [1 , np.nan , np.nan , np.nan , 5]
}
# np.nan is not a number(nan)
df = pd.DataFrame(data)
```

```python
df
```

```python
df.isna()  #Gives False for NA values such as None , and np.NaN. It doe snot consider "" , np.if
```

```python
# Find number of Null values
df.isna().sum()  # Use ful for Large DataSets
```

```python
# To check whether null value exist
df.isna().any()
```

### Removing Missing Data


```python
df
```

### Filling the missing Data

```python
df
```

```python
df.fillna(0)  ## Column wise
```

```python
values = { 'A':0 , 'B': 100 , 'C' : 300 , 'D' : 400}
df.fillna(value = values)  # inplace = True to change the df
```

```python
df
```

```python
## To fill mean values in place of null value

df.fillna(df.mean())
```
